```python
block_size = 10 # one day
n_boot = 100000

# Wiener process model  ## ONLY UNCOMMENT ONE BLOCK OF 3 AT A TIME (I.E., EITHER E, I, OR BETA)
metric_a = {
    # "mean_bias": e_error,         
    # "median_bias": e_med_bias,
    # "variance": e_var

    # "mean_bias": i_error,         
    # "median_bias": i_med_bias,
    # "variance": i_var

    "mean_bias": np.abs(beta_error),         
    "median_bias": np.abs(beta_med_bias),
    "variance": beta_var
}

# Sum-of-sigmoid model
metric_b = {
    # "mean_bias": np.abs(e_error_sos_forecast),       
    # "median_bias": np.abs(e_med_bias_sos_forecast),  
    # "variance": e_var_sos_forecast          
    
    # "mean_bias": np.abs(i_error_sos_forecast),       
    # "median_bias": np.abs(i_med_bias_sos_forecast),  
    # "variance": i_var_sos_forecast           

    "mean_bias": np.abs(beta_error_sos_forecast),   
    "median_bias": np.abs(beta_med_bias_sos_forecast),  
    "variance": beta_var_sos_forecast           
}

def moving_block_bootstrap_pval(a, b, block_size=3, n_boot=5000):
    n = len(a)
    starts = np.arange(0, n - block_size + 1)            # overlapping block starts
    n_blocks = len(starts)

    # make overlapping blocks
    a_blocks = np.array([a[s:s+block_size] for s in starts])
    b_blocks = np.array([b[s:s+block_size] for s in starts])

    boot_diffs = np.zeros(n_boot)
    for k in range(n_boot):
        idx = np.random.randint(0, n_blocks, size=n_blocks)
        a_sample = a_blocks[idx].ravel()[:n]             # flatten then trim
        b_sample = b_blocks[idx].ravel()[:n]
        boot_diffs[k] = np.mean(a_sample - b_sample)

    p_value = np.mean(boot_diffs <= 0)   # p small => B better

    # plt.figure()
    # plt.plot(boot_diffs)

    return p_value


pvals = {}
for metric in ["mean_bias","median_bias","variance"]:
    a_series = metric_a[metric]
    b_series = metric_b[metric]
    pvals[metric] = moving_block_bootstrap_pval(a_series, b_series,
                                                block_size=block_size, n_boot=n_boot)

print("Block bootstrap p-values (1-day blocks):")
for metric, p in pvals.items():
    print(f"{metric}: {p:.4g}")
```
